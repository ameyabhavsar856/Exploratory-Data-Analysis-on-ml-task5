```python

# Elevate Labs - Data Analyst Internship
# Task 5: Exploratory Data Analysis (EDA) on Titanic Dataset

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)


df = sns.load_dataset("titanic")


print("First 5 rows of the dataset:")
print(df.head())


print("\n--- Dataset Info ---")
df.info()

print("\n--- Summary Statistics (Numerical Features) ---")
print(df.describe())

print("\n--- Summary Statistics (Categorical Features) ---")
print(df.describe(include=["object", "category"]))

print("\n--- Missing Value Counts ---")
print(df.isnull().sum())

print("\n--- Categorical Value Counts: Survived ---")
print(df["survived"].value_counts(normalize=True) * 100)

print("\n--- Categorical Value Counts: Class ---")
print(df["pclass"].value_counts())

print("\n--- Categorical Value Counts: Sex ---")
print(df["sex"].value_counts())



plt.figure(figsize=(8, 5))
sns.histplot(df["age"].dropna(), kde=True, bins=30, color="teal")
plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()



plt.figure(figsize=(8, 4))
sns.boxplot(x=df["fare"], color="salmon")
plt.title("Boxplot of Ticket Fare")
plt.xlabel("Fare")
plt.show()

# Visual 3: Bivariate Analysis - Survival Rate by Sex (Barplot)
plt.figure(figsize=(6, 5))
sns.barplot(x="sex", y="survived", data=df, errorbar=None, palette="muted")
plt.title("Survival Rate by Sex")
plt.ylabel("Survival Probability")
plt.show()

plt.figure(figsize=(6, 5))
sns.barplot(x="pclass", y="survived", data=df, errorbar=None, palette="Blues_d")
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Ticket Class (1 = 1st, 2 = 2nd, 3 = 3rd)")
plt.ylabel("Survival Probability")
plt.show()


plt.figure(figsize=(9, 6))
sns.scatterplot(
    x="age", y="fare", hue="survived", style="survived", alpha=0.7, data=df
)
plt.title("Age vs Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()


plt.figure(figsize=(8, 6))
numeric_cols = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix Heatmap")
plt.show()


sns.pairplot(
    df[["survived", "pclass", "age", "fare"]].dropna(),
    hue="survived",
    diag_kind="kde",
)
plt.show()
```
